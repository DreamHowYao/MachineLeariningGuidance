### 2. 可视化ndarray格式的图片数据

```python
# 2.1 加载数据
data = pd.read_csv('./data/number.csv')
data.head()
```

```python
idx = data.iloc[3,1:].to_numpy()
idx_reshape = idx.reshape(28, 28)
plt.imshow(idx_reshape,cmap='gray')
plt.plot()
```

```python
idx = data.iloc[89,1:].to_numpy()
idx_reshape = idx.reshape(28, 28)
plt.imshow(idx_reshape,cmap='gray')
plt.plot()
```
