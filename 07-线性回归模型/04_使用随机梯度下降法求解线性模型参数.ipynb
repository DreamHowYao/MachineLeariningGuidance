```python
from sklearn.datasets import fetch_openml
import pandas as pd

# 1.从OpenML加载数据集,指定名称为 "house_prices"
# 设置 as_frame=True 可以直接获得pandas的DataFrame,方便查看和操作
ames = fetch_openml(name="house_prices", as_frame=True) ## 返回Bunch对象
```

```python
# 1. 数据预处理
X: pd.DataFrame = ames.data
Y = ames.target

## 1.1 特征筛选
X = X[['MSSubClass','LotArea','OverallCond']]

## 1.2 去除重复项
keep_mask = ~X.duplicated(keep='first')
X = X[keep_mask]
Y = Y[keep_mask]
X.head(3)
```

```python
# 3. 划分数据集
from sklearn.model_selection import train_test_split
# 指定随机数种子,方便复现
SEED = 42

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=SEED)
```

```python
# 5. 实例化线性回归模型
from sklearn.linear_model import SGDRegressor
sgd_estimator = SGDRegressor(fit_intercept=True,loss='squared_error')
# 6. 模型训练
sgd_estimator.fit(X_train, y_train)
# 6. 模型预测
y_pred = sgd_estimator.predict(X_test)
# 7. 模型评估
from sklearn.metrics import mean_squared_error, mean_absolute_error
print(f'mean_absolute_error:{mean_absolute_error(y_test, y_pred)}')
print(f'mean_squared_error:{mean_squared_error(y_test, y_pred)}')
```
