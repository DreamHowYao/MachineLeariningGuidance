```python
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
```

```python
# =====================================
# 训练数据
# =====================================

train_x = np.array([
    [1, 0],
    [1, 1]
])

# 连续值标签（回归）
train_y = np.array([
    100,
    0
])

# 测试数据
sample_x = np.array([
    [0, 0],
    [1, 1]
])
```

```python
"""
KNN回归器的工作原理：
1. 计算距离：对于每个要预测的新样本,计算它到所有训练样本的距离
2. 找邻居：选出距离最近的 n_neighbors 个训练样本
3. 加权计算：根据weights参数决定如何计算最终结果
   - "uniform"：所有邻居平等,直接取平均值
   - "distance"：距离越近的邻居权重越大
4. 预测结果：用加权平均值作为预测值
"""
uniform_regressor = KNeighborsRegressor(n_neighbors=2,weights="uniform",metric="minkowski",p=2)

distance_regressor = KNeighborsRegressor(n_neighbors=2,weights="distance",metric="minkowski",p=2)

# 只存储数据,不训练模型
uniform_regressor.fit(train_x, train_y)
distance_regressor.fit(train_x, train_y)

print("均匀权重预测结果：", uniform_regressor.predict(sample_x))
print("距离权重预测结果：", distance_regressor.predict(sample_x))
```

```python
def get_weights(dist):
    """
    自定义权重函数：距离的平方的倒数
    这样距离越近,权重越大

    :param：
        dist：距离数组,每一行是一个测试样本到所有训练样本的距离
    :return：
        权重数组

    - 如果距离为0,权重无穷大
    - 距离越大,权重越小(1 / d ^ 2)
    """
    with np.errstate(divide="ignore"):
            dist = 1.0 / dist ** 2
            inf_mask = np.isinf(dist)
            inf_row = np.any(inf_mask, axis=1)
            dist[inf_row] = inf_mask[inf_row]
    return dist

my_regressor = KNeighborsRegressor(n_neighbors=2,weights=get_weights).fit(train_x, train_y)
```

```python
print("自定义权重预测结果：", my_regressor.predict(sample_x))
```
