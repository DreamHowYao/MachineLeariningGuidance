### 使用3σ原则检测数据中的离群点
原理:如果特征数据近似服从正态分布,其大部分特征值应该在$(\mu \pm 3 \sigma)$之内

```python
from pandas.core.interchange.dataframe_protocol import DataFrame
# 加载内置的糖尿病数据集（包含442个样本,10个特征）
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
```

```python
# 1.加载数据集
data = load_diabetes(as_frame=True)
type(data)
```

```python
# 2. 获取对应的DataFrame格式
print("data.data:",type(data.data))
X: pd.DataFrame = data.data
Y: pd.DataFrame = data.target
```

```python
# 2.1 查看数据
X.head(5)
```

```python
Y.head(5)
```

```python
# 2.2 标准化
scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)
X_scaled.head(5)
```

```python
# 3. 使用检测数据中的离群点(异常值)——3σ原则
outlier_mask = (
    np.abs(X_scaled)
    > 3
).any(axis=1)
# 查看异常数据,并决定剔除或进行特征转换
X[outlier_mask]
```
