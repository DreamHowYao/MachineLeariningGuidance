```python
# 获取标准化API
from sklearn.preprocessing import StandardScaler
```

```python
# 1.构造数据
data = [[90, 2, 10, 40],
        [60, 4, 15, 45],
        [75, 3, 13, 46]]
```

```python
# 2.获取实例对象
transformer = StandardScaler()
```

```python
# 3.标准化数据
print(f"原始数据:\r\n{data}")
data_preprocess = transformer.fit_transform(data)
print(f"归一化后的数据:\r\n{data_preprocess}")

# 3.1 等价数据预处理步骤
# fit = transformer.fit(data)
# data_preprocess = transformer.transform(data)
```
