```python
# 归一化API
from sklearn.preprocessing import MinMaxScaler
```

```python
# 1.构造数据
data = [[90, 2, 10, 40],
        [60, 4, 15, 45],
        [75, 3, 13, 46]]
```

```python
# 2.获取实例对象
transformer = MinMaxScaler(feature_range=(2, 3))
# 3.归一化
"""
1. fit()方法
    - 作用:从训练数据学习参数（如均值、方差、主成分等）,
2. transform()方法
    - 作用:使用已学习的参数转换训练集、验证集或测试集
3. fit_transform()方法
    - 作用:在训练集上一步完成学习和转换,通常更简洁,有时也更高效
"""
transformer.fit(data) # 计算数据中的最大-最小值以便于后续的归一化
n_data = transformer.transform(data) # 归一化

# 3.1 使用fit_transform()方法等同于上面两步
# transformer.fit_transform(data)
```

```python
# 3.归一化
print(f"原始数据:\r\n{data}")
print(f"归一化后的数据:\r\n{n_data}")
```
